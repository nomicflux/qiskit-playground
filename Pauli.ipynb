```python
from qiskit.quantum_info import Pauli
```

```python
P = Pauli("-iXYZ")
print(f'P[0] = {repr(P[0])}')
print(f'P[1] = {repr(P[1])}')
print(f'P[2] = {repr(P[2])}')
print(f'P[:] = {repr(P[:])}')
print(f'P[::-1] = {repr(P[::-1])}')
```

```python
from qiskit import transpile
```

```python
from qiskit.circuit.library import MCXGate
from qiskit import QuantumCircuit
```

```python
mcx = MCXGate(4)
```

```python
sc = QuantumCircuit(5)
sc.append(mcx, [0,1,4,2,3])
sc_tqc = transpile(sc, basis_gates=['u', 'cx'])
print(f'No aux: {sc_tqc.count_ops()}')

lc = QuantumCircuit(10)
lc.append(mcx, [0,1,4,2,3])
lc_tqc = transpile(lc, basis_gates=['u', 'cx'])
print(f'With aux: {lc_tqc.count_ops()}')
```
