```python
from qiskit import QuantumCircuit
from qiskit.visualization import circuit_drawer
```

```python
qc = QuantumCircuit(4)
qc.h(0)
qc.cx(0,1)
qc.cx(0,2)
qc.cx(0,3)
circuit_drawer(qc, output='mpl')
```

```python
from qiskit_ibm_runtime import QiskitRuntimeService
from qiskit.transpiler import generate_preset_pass_manager
from qiskit.visualization import plot_gate_map
```

```python
service = QiskitRuntimeService()
```

```python
backends = service.backends()
print(f'Backends: {backends}')
```

```python
backend_fez = service.backend('ibm_fez')
backend_marrakesh = service.backend('ibm_marrakesh')
backend_torino = service.backend('ibm_torino')
```

```python
plot_gate_map(backend_fez, plot_directed=True)
```

```python
plot_gate_map(backend_marrakesh, plot_directed=True)
```

```python
plot_gate_map(backend_torino, plot_directed=True)
```

```python
from qiskit.visualization import plot_circuit_layout
pass_manager = generate_preset_pass_manager(backend=backend_fez, optimization_level = 0)
qc_fez_0 = pass_manager.run(qc)
print(f'Depth: {qc_fez_0.depth()}')
plot_circuit_layout(qc_fez_0, backend_fez)
```

```python
circuit_drawer(qc_fez_0, idle_wires=False, output='mpl')
```

```python
pass_manager_torino = generate_preset_pass_manager(backend=backend_torino, optimization_level = 0)
qc_torino_0 = pass_manager_torino.run(qc)
print(f'Depth: {qc_torino_0.depth()}')
```

```python
plot_circuit_layout(qc_torino_0, backend_torino)
```

```python
circuit_drawer(qc_torino_0, idle_wires=False, output='mpl')
```

```python
pass_manager_fez_3 = generate_preset_pass_manager(backend=backend_fez, optimization_level=3)
qc_fez_3 = pass_manager_fez_3.run(qc)
print(f'Depth: {qc_fez_3.depth()}')
```

```python
print(f'Ops count: {qc_fez_3.count_ops()}')
```

```python
plot_circuit_layout(qc_fez_3, backend_fez)
```

```python
circuit_drawer(qc_fez_3, idle_wires=False, output='mpl')
```

```python
pass_manager_torino_3 = generate_preset_pass_manager(backend=backend_torino, optimization_level=3)
qc_torino_3 = pass_manager_torino_3.run(qc)
print(f'Depth: {qc_torino_3.depth()}')
print(f'Ops count; {qc_torino_3.count_ops()}')
```

```python
plot_circuit_layout(qc_torino_3, backend_torino)
```

```python
circuit_drawer(qc_torino_3, idle_wires=False, output='mpl')
```

```python
ibm_fez_coupling_map = backend_fez.configuration().coupling_map
ibm_fez_coupling_map
```

```python
pass_manager_none_fez_3 = generate_preset_pass_manager(backend=None, optimization_level=3, coupling_map=ibm_fez_coupling_map)
qc_custom = pass_manager_none_fez_3.run(qc)
circuit_drawer(qc_custom, idle_wires=False, output='mpl')
```

```python
custom_linear_topology = [[0,1], [1,2], [2,3], [3,4]]
pass_manager_custom_linear_3 = generate_preset_pass_manager(backend=None, optimization_level=3, coupling_map=custom_linear_topology)
qc_custom = pass_manager_custom_linear_3.run(qc)
circuit_drawer(qc_custom, idle_wires=False, output='mpl')
```

```python
from qiskit.transpiler import PassManager, CouplingMap
from qiskit.transpiler.passes import TrivialLayout, BasicSwap
```

```python
pm = PassManager()
trivial = TrivialLayout(CouplingMap(ibm_fez_coupling_map))
pm.append(trivial)
tv_qc = pm.run(qc)
circuit_drawer(tv_qc, idle_wires=False, output='mpl')
```

```python
basic_swap = BasicSwap(CouplingMap(ibm_fez_coupling_map))
pm = PassManager(basic_swap)
new_qc = pm.run(qc)
circuit_drawer(new_qc, idle_wires=False, output='mpl')
```

```python
qc2 = QuantumCircuit(4)
qc2.h(0)
qc2.cx(0,1)
qc2.barrier()
qc2.cx(0,2)
qc2.cx(0,3)
qc2.barrier()
qc2.cz(3,0)
qc2.h(0)
qc2.measure_all()
circuit_drawer(qc2, output='mpl')
```

```python
style = {'backgroundcolor': 'lightblue', 'gatefacecolor': 'white', 'gatetextcolor': 'black', 'fontsize': 9}
circuit_drawer(qc2, style=style, output='mpl')
```

```python
from qiskit.converters import circuit_to_dag
from qiskit.visualization import dag_drawer
```

```python
dag = circuit_to_dag(qc2)
dag_drawer(dag)
```
